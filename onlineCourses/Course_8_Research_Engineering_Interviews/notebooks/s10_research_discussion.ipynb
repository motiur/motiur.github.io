### 2. State your result with its uncertainty

```python
import numpy as np
from scipy import stats

def mean_ci(runs):
    runs = np.asarray(runs, dtype=float)
    se = runs.std(ddof=1) / np.sqrt(len(runs))
    half = stats.t.ppf(0.975, len(runs) - 1) * se
    return runs.mean(), half

mine = [70.1, 71.9, 71.0, 72.4, 70.6]         # TODO: your method's seeds
base = [69.8, 70.9, 70.2, 71.5, 70.0]         # TODO: the baseline's seeds
for name, r in [("method", mine), ("baseline", base)]:
    m, h = mean_ci(r)
    print(f"{name}: {m:.2f} +/- {h:.2f}")
t, p = stats.ttest_ind(mine, base)
print(f"difference {np.mean(mine) - np.mean(base):.2f}, two-sample t-test p = {p:.3f}")
```

If the intervals overlap heavily, the honest sentence is: "we found no clear difference at five runs".
