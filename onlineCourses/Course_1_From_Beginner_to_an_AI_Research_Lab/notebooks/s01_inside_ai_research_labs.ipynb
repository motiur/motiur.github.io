### 1. Five labs (worksheet)
Start from a paper or tool you have heard of, find its authors, and open their lab page. Include at least one university lab, one industry lab and one open community. Replace the example rows.

```python
import pandas as pd
labs = pd.DataFrame([
    {"lab": "<university lab>", "kind": "university", "works_on": "<one line from its three newest abstracts>", "skills_shown": "<e.g. PyTorch, statistics, a domain>"},
    {"lab": "<industry lab>", "kind": "industry", "works_on": "", "skills_shown": ""},
    {"lab": "<open community>", "kind": "open community", "works_on": "", "skills_shown": ""},
    {"lab": "<lab 4>", "kind": "", "works_on": "", "skills_shown": ""},
    {"lab": "<lab 5>", "kind": "", "works_on": "", "skills_shown": ""},
])
def check_labs(df):
    kinds = set(df.kind)
    missing = [k for k in ["university", "industry", "open community"] if k not in kinds]
    return "ok" if len(df) >= 5 and not missing else f"still needed: {missing or 'five labs'}"
print(labs.to_string(index=False)); print(check_labs(labs))
```

### 2. Two postings (worksheet)
Copy every requirement from two real postings (a lab's 'join us' page, a careers page, or a fellowship such as a predoctoral or scholars program). One requirement per line.

```python
posting_1 = """PyTorch experience
Research experience
Publications or reports
Strong letters of recommendation"""
posting_2 = """Python and PyTorch
Experience running ML experiments
A degree in a quantitative field
Clear written communication"""
requirements = sorted({line.strip() for p in (posting_1, posting_2) for line in p.splitlines() if line.strip()})
print(len(requirements), "distinct requirements")
```

### 3. The skills-gap table
For each requirement: what evidence do you have today, and which part of this course closes the gap? The helper suggests sessions from keywords; correct it by hand.

```python
SESSIONS = {"math": "Session 2", "python": "Session 3", "pytorch": "Sessions 3, 5, 6", "experiment": "Sessions 8, 9", "research experience": "Sessions 8-11 capstone",
            "publication": "Session 11 paper", "report": "Session 11 paper", "letter": "Mentor by session 8", "written": "Sessions 7, 11", "communication": "Sessions 7, 11, 12",
            "paper": "Session 7", "statistic": "Session 9", "degree": "outside this course"}
def closed_by(req):
    hits = [v for k, v in SESSIONS.items() if k in req.lower()]
    return "; ".join(dict.fromkeys(hits)) or "TODO: decide"
gap = pd.DataFrame({"requirement": requirements, "my_evidence_today": "none", "closed_by": [closed_by(r) for r in requirements]})
print(gap.to_string(index=False))
assert (gap.closed_by != "").all()
```

Most first-week tables show 'none' in the evidence column. That is the starting point, and the course exists to fill it. Come back to this table after sessions 6, 9 and 12 and update the middle column.
