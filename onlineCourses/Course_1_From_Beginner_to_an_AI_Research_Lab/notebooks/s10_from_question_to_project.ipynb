### 2. Score each question from 1 to 3 on four tests
- Answerable in 3 months.
- Fits my compute.
- Interesting either way (a negative result would still be worth reporting).
- Clear baseline and metric.

The scores below are examples. Replace them with yours.

```python
scores = pd.DataFrame({"id": ["Q1", "Q2", "Q3", "Q4", "Q5"],
                       "answerable_in_3_months": [3, 2, 2, 1, 3], "fits_my_compute": [3, 2, 3, 1, 3],
                       "interesting_either_way": [2, 3, 2, 3, 1], "clear_baseline_and_metric": [3, 2, 1, 2, 3]})
scores["total"] = scores.drop(columns="id").sum(axis=1)
scores["weakest_test"] = scores.drop(columns=["id", "total"]).idxmin(axis=1)
print(scores.sort_values("total", ascending=False).to_string(index=False))
best = scores.sort_values("total", ascending=False).iloc[0]
print("keep:", best.id, "| its weakest point to fix in the proposal:", best.weakest_test)
```

The highest-scoring question is usually narrower than the one you liked most at first. That is the right direction. A score of 1 on any test is a warning even when the total is high.

### 3. Is it a research question yet?
A question needs something to compare against and a way to measure. This helper flags the usual problems in the wording.

```python
def critique(q):
    issues = []
    low = q.lower()
    if not q.strip().endswith("?"): issues.append("not phrased as a question")
    if any(w in low for w in ["build a system", "develop a", "create a", "make a"]): issues.append("describes a product, not a question: what would you compare it with?")
    if not any(w in low for w in ["than", "compared", "versus", "vs", "more", "less", "better", "worse", "does", "how much", "whether"]): issues.append("no comparison in sight")
    if len(q.split()) > 40: issues.append("too long: probably two questions")
    return issues or ["looks like a testable question"]
for q in ["Build a system that detects dengue outbreaks.",
          "Does adding a rotary position embedding lower validation loss compared with a learned one on a 10M-parameter Bangla model at equal steps?"]:
    print(q, "\n   ->", critique(q))
assert critique("Build a system that detects dengue outbreaks.") != ["looks like a testable question"]
```

### 4. The one-page proposal

### 5. Twelve weeks, with a checkpoint that can stop the project
Plan backwards from the write-up. The week-4 result is a decision point: continue, narrow, or switch.

```python
plan = pd.DataFrame([
    {"weeks": "1-2", "work": "baseline reproduced in my own code; evaluation fixed", "deliverable": "baseline number over 3 seeds"},
    {"weeks": "3-4", "work": "cheapest experiment that could prove the idea wrong", "deliverable": "CHECKPOINT: one plot, go / narrow / switch"},
    {"weeks": "5-7", "work": "main experiments over 5 seeds", "deliverable": "results table with intervals"},
    {"weeks": "8-9", "work": "ablations; fair baseline tuning", "deliverable": "ablation table"},
    {"weeks": "10-11", "work": "writing; figures; missing runs", "deliverable": "4-page draft"},
    {"weeks": "12", "work": "review by a classmate; revision; code release", "deliverable": "paper + repository"},
])
print(plan.to_string(index=False))
def compute_check(runs, hours_per_run, weeks=12, quota_per_week=30, safety=2.0):
    need = runs * hours_per_run * safety; have = weeks * quota_per_week
    return f"planned {runs * hours_per_run:.0f} GPU hours, with a safety factor of {safety:.0f}: {need:.0f} | available {have} | {'fits' if need <= have else 'does NOT fit: cut scope now'}"
print(compute_check(runs=40, hours_per_run=3)); print(compute_check(runs=40, hours_per_run=6))
```

The safety factor of 2 is there because the first plan always forgets the reruns after a bug.
