### 1. Size arithmetic
Weights take `parameters x bytes per parameter`.

```python
def size_mb(params, bytes_per_param=4):
    return params * bytes_per_param / 1e6

for name, p in [("10M model", 10e6), ("1B model", 1e9), ("3B model", 3e9)]:
    print(f"{name}: {size_mb(p):,.0f} MB at 32-bit, {size_mb(p, 2):,.0f} MB at 16-bit")
```
