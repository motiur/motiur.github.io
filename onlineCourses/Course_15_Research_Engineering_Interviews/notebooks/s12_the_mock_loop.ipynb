### 1. Schedule for a group of four
Each person is a candidate for rounds A-D and interviews the others in between. Rounds are 40 minutes with 10-minute breaks.

```python
people = ["P1", "P2", "P3", "P4"]                     # TODO: your names
rounds = ["A coding", "B ML coding", "C fundamentals", "D research"]
slot, seen = 1, set()
for r, rnd in enumerate(rounds):
    for candidates, interviewers in [(people[:2], people[2:]), (people[2:], people[:2])]:
        pairs = [(c, interviewers[(i + r) % 2]) for i, c in enumerate(candidates)]
        print(f"slot {slot}: " + "; ".join(f"{c} is interviewed by {i} ({rnd})" for c, i in pairs))
        seen.update((c, rnd) for c, _ in pairs)
        slot += 1
print("every person sits every round:", len(seen) == 16)
```

Each slot has two interviews running in parallel: two people are candidates and the other two interview them. Over eight slots, everyone sits all four rounds and interviews four times.
