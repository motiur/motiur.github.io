### 1. Shortlist three papers (worksheet)
Each candidate needs public code or enough detail to re-implement, one result on a small model or dataset, and a clear metric.

```python
import pandas as pd
shortlist = pd.DataFrame([
    {"paper": "<title 1>", "year": None, "code_public": None, "small_scale_result": "<which table or figure>", "metric": "", "est_gpu_hours": None},
    {"paper": "<title 2>", "year": None, "code_public": None, "small_scale_result": "", "metric": "", "est_gpu_hours": None},
    {"paper": "<title 3>", "year": None, "code_public": None, "small_scale_result": "", "metric": "", "est_gpu_hours": None},
])
def suitable(row, quota_hours=30):
    return bool(row.code_public) and bool(row.small_scale_result) and bool(row.metric) and row.est_gpu_hours is not None and row.est_gpu_hours * 7 <= quota_hours
print(shortlist.to_string(index=False))
print("suitable for a capstone (5 seeds + baseline + ablation must fit the quota):", [suitable(r) for r in shortlist.itertuples()])
```

The factor of 7 is deliberate: one run of the paper's result is not a reproduction. You need about five seeds, a baseline and an ablation.

### 2. Second pass: claims and evidence
For each claim in the abstract, write the table or figure that supports it. A claim with nothing in the last columns is the most interesting line on the page.

```python
claims = pd.DataFrame([
    {"claim_in_abstract": "<e.g. 'our method improves accuracy by 2 points on X'>", "evidence": "<Table 2, row 3>", "n_seeds": None, "baseline": "", "my_confidence_1_to_3": None},
    {"claim_in_abstract": "", "evidence": "", "n_seeds": None, "baseline": "", "my_confidence_1_to_3": None},
])
def unsupported(df): return df[(df.claim_in_abstract != "") & ((df.evidence == "") | df.n_seeds.isna())]
print(claims.to_string(index=False)); print("claims with missing evidence or unknown seeds:", len(unsupported(claims)))
```

### 3. Third pass: every detail a reproduction needs
Mark where each detail is found: `stated` in the paper, in the `appendix`, in the `code only`, or `missing`.

```python
DETAILS = ["dataset and version", "preprocessing", "train/validation/test split", "model size and architecture", "initialization", "optimizer and its settings",
           "learning-rate schedule", "batch size", "number of steps or epochs", "number of seeds", "how the metric is computed", "hardware and run time"]
details = pd.DataFrame({"detail": DETAILS, "value": "", "found_in": "TODO"})          # found_in: stated / appendix / code only / missing
def summary(df): return df.found_in.value_counts().to_dict()
print(details.to_string(index=False)); print(summary(details))
```

Expect several entries marked 'code only' or 'missing'. That list is the first draft of next session's reproduction plan and of its deviations log.

### 4. The one-page summary

```python
SUMMARY = """Paper, venue and year:
Problem (one sentence):
Main claim (one sentence, with the number):
Evidence (which table or figure, how many seeds, what baseline):
Limitations the authors state:
Limitations I see:
The one result I could reproduce, and its compute:
A question I would test next:
"""
open("paper_summary.md", "w").write(SUMMARY); print(SUMMARY)
def word_count(path): return len(open(path).read().split())
print("words so far:", word_count("paper_summary.md"), "(aim for under 400 when filled in)")
```

### 5. The literature map
Five papers that yours cites and five that cite it, with one line each on how they relate. Then look at the shape.

```python
lit = pd.DataFrame([{"paper": f"<cited {i}>", "direction": "cited by mine", "relation": "<builds on / compares with / uses data from / disagrees with>"} for i in range(1, 6)] +
                   [{"paper": f"<citing {i}>", "direction": "cites mine", "relation": ""} for i in range(1, 6)])
print(lit.groupby("direction").size().to_dict())
# Where to find 'cites mine': the paper's page on Semantic Scholar or Google Scholar, under 'cited by'.
# Read the map: which later paper tested the same claim? Did it confirm it?
```
