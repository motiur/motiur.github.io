### The six problems
1. `two_sum(nums, target)`: indices of two numbers that add to target, or None.
2. `group_anagrams(words)`: list of groups, each sorted, groups sorted.
3. `valid_brackets(s)`: are `()[]{}` balanced and properly nested?
4. `merge_intervals(intervals)`: merge overlapping `[start, end]` pairs.
5. `shortest_path(grid)`: fewest steps from top-left to bottom-right through 0 cells (4 directions), or -1.
6. `top_k_frequent(words, k)`: the k most frequent words, ties broken by larger word first.

```python
def run_tests(two_sum, group_anagrams, valid_brackets, merge_intervals, shortest_path, top_k_frequent):
    assert sorted(two_sum([2, 7, 11, 15], 9)) == [0, 1] and two_sum([1, 2], 7) is None
    assert group_anagrams(["eat", "tea", "tan", "ate", "nat", "bat"]) == [["ate", "eat", "tea"], ["bat"], ["nat", "tan"]]
    assert valid_brackets("{[()]}") and not valid_brackets("([)]") and not valid_brackets("(") and valid_brackets("")
    assert merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]) == [[1, 6], [8, 10], [15, 18]]
    assert merge_intervals([[1, 4], [4, 5]]) == [[1, 5]] and merge_intervals([]) == []
    assert shortest_path([[0, 0, 0], [1, 1, 0], [0, 0, 0]]) == 4 and shortest_path([[0, 1], [1, 0]]) == -1
    assert top_k_frequent("a b a c b a d".split(), 2) == ["a", "b"]
    print("all six pass")
```

## Reference solutions
Try every problem under the timer first. Then run this section: it defines working versions and runs the same tests.

```python
import heapq
from collections import Counter, defaultdict, deque

def two_sum(nums, target):                       # O(n) time, O(n) space
    seen = {}
    for i, x in enumerate(nums):
        if target - x in seen:
            return [seen[target - x], i]
        seen[x] = i
    return None

def group_anagrams(words):                       # O(n k log k)
    groups = defaultdict(list)
    for w in words:
        groups["".join(sorted(w))].append(w)
    return sorted(sorted(g) for g in groups.values())

def valid_brackets(s):                           # O(n)
    pairs, stack = {")": "(", "]": "[", "}": "{"}, []
    for ch in s:
        if ch in pairs:
            if not stack or stack.pop() != pairs[ch]:
                return False
        else:
            stack.append(ch)
    return not stack

def merge_intervals(intervals):                  # O(n log n)
    out = []
    for start, end in sorted(intervals):
        if out and start <= out[-1][1]:
            out[-1][1] = max(out[-1][1], end)
        else:
            out.append([start, end])
    return out

def shortest_path(grid):                         # BFS, O(rows * cols)
    rows, cols = len(grid), len(grid[0])
    if grid[0][0] or grid[-1][-1]:
        return -1
    queue, seen = deque([(0, 0, 0)]), {(0, 0)}
    while queue:
        r, c, d = queue.popleft()
        if (r, c) == (rows - 1, cols - 1):
            return d
        for nr, nc in ((r + 1, c), (r - 1, c), (r, c + 1), (r, c - 1)):
            if 0 <= nr < rows and 0 <= nc < cols and not grid[nr][nc] and (nr, nc) not in seen:
                seen.add((nr, nc)); queue.append((nr, nc, d + 1))
    return -1

def top_k_frequent(words, k):                    # O(n log k)
    counts = Counter(words)
    return heapq.nlargest(k, counts, key=lambda w: (counts[w], w))

run_tests(two_sum, group_anagrams, valid_brackets, merge_intervals, shortest_path, top_k_frequent)
```
