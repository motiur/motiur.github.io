### Worked numbers to know

```python
# Bayes: prevalence 1%, sensitivity 99%, specificity 95%
prev, sens, spec = 0.01, 0.99, 0.95
p_pos = prev * sens + (1 - prev) * (1 - spec)
print("P(disease | positive) =", round(prev * sens / p_pos, 3))

# Variance of a mean of n draws is sigma^2 / n: check by simulation
import numpy as np
rng = np.random.default_rng(0)
means = rng.normal(0, 2, (20000, 25)).mean(1)
print("simulated variance of the mean:", round(means.var(), 3), "| sigma^2 / n =", 4 / 25)
```
